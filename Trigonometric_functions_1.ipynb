```python
angle_degree = 30
angle_radian = angle_degree * math.pi / 180
print("angle_radian : ",angle_radian)
```

직교 좌표를 극좌표로 전환

```python
x, y = 1, 2
```

```python
r = math.sqrt(x**2 + y**2)
theta_radian = math.atan2(y, x)
print(r, theta_radian)
```

극좌표를 직교 좌표로 전환

```python
r,angle = 10,30
```

```python
theta_radian = math.radians(angle)
x = r * math.cos(theta_radian)
y = r * math.sin(theta_radian)
print(x, y)
```
